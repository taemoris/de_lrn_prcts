```python
from pyspark.sql import SparkSession

spark = (
    SparkSession
        .builder
        .appName("spark-local-app")
        .master("local[*]")
        .getOrCreate()
)
```

```python
import datetime

orders = [
    {
        "order_id": 1,
        "customer_name": "Tembulat",
        "order_date": datetime.datetime(2026, 12, 12, 1, 1, 1),
        "items": {
            "product_name": "Iphone 12 512 Gb",
            "quantity": 1,
            "price": 32000.2334
        }
    },
    {
            "order_id": 2,
            "customer_name": "Olga",
            "order_date": datetime.datetime(2026, 10, 3, 12, 10, 1),
            "items": {
                "product_name": "Soap Duru",
                "quantity": 3,
                "price": 300.43
            }
        },
    {
            "order_id": 3,
            "customer_name": "Thomas",
            "order_date": datetime.datetime(2026, 3, 12, 13, 12, 1),
            "items": {
                "product_name": "Tecno Buds 3",
                "quantity": 1,
                "price": 1200.344
            }
    },
]
```

```python
from pyspark.sql.types import IntegerType, DateType, StringType, StructField, StructType, DoubleType

ITEMS = StructType([
    StructField("product_name", StringType(), nullable=False),
    StructField("quantity", IntegerType(), nullable=False),
    StructField("price", DoubleType(), nullable=False)
])

ORDER_SCHEMA = StructType([
    StructField("order_id", IntegerType(), nullable=False),
    StructField("customer_name", StringType(), nullable=False),
    StructField("order_date", DateType(), nullable=False),
    StructField("items", ITEMS, nullable=False)
])

df_orders = spark.createDataFrame(orders, schema=ORDER_SCHEMA)
df_orders.printSchema()
```

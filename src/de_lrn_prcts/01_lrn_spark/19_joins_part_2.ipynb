```python
df_users = spark.createDataFrame(data_users)
df_users2 = spark.createDataFrame(data_users_2)
```

```python
(
    df_users
        .join(df_users2, "email", "full")
        .show()
)
```

```python
from pyspark.sql.functions import coalesce

(
    df_users
        .join(df_users2, "email", "full")
        .select(
            coalesce(df_users["email"], df_users2["email"]).alias("email"),
            coalesce(df_users["first_name"], df_users2["first_name"]).alias("first_name"),
            coalesce(df_users["gender"], df_users2["gender"]).alias("gender"),
            coalesce(df_users["last_name"], df_users2["last_name"]).alias("last_name"),
            coalesce(df_users["ip_address"], df_users2["ip_address"]).alias("ip_address"),
        )
        .show()
)
```

```python
from pyspark.sql import SparkSession

spark = (
    SparkSession
        .builder
        .appName("spark-local-app")
        .master("local[*]")
        .getOrCreate()
)
```

```python
employees = [
    (1, "Bob", "Tiger", 12432.343, 10, "united States", "+1 2343434 34", "43 54 5465"),
    (2, "Alice", "Smith", 35000.50, 5, "CANADA", "+1 555 123 4567", "123 Main St"),
    (3, "John", "Doe", 28900.12, None, "UK", "+44 7700 900123", "10 Downing St"),
    (4, "Maria", "Gonzalez", 42000.75, 22, "MExiCO", "+52 55 1234 5678", "Av. Reforma 123"),
    (5, "Ken", "Tanaka", 39500.90, None, "Japan", "+81 3 1234 5678", "Tokyo, Chuo-ku"),
    (6, "Olga", "Petrova", 31000.25, '', "RUSSIa", "+7 999 123 45 67", "Moscow, Tverskaya 1"),
    (7, "Liam", "Brown", 27500.80, 5, "Australia", "+61 2 1234 5678", "Sydney, George St"),
    (8, "Emma", "Lee", 45000.60, None, "Singapore", "+65 6123 4567", "Orchard Road 10"),
    (9, "Diego", "Rodriguez", 33000.40, 5, "SPain", "+34 91 123 45 67", "Madrid, Gran Via"),
    (10, "Sophie", "Martin", 38000.30, 45, "France", "+33 1 23 45 67 89", "Paris, Champs-Élysées")
]
```

```python
df_employees = spark.createDataFrame(
    employees,
    schema="""
    id INT,
    first_name STRING,
    last_name STRING,
    salary DOUBLE,
    bonus STRING,
    country STRING,
    phone_number STRING,
    address STRING
    """
)
```

```python
df_employees.show()
```

```python
from pyspark.sql.functions import col

(
    df_employees
        .withColumn("bonus_changed", col("bonus").try_cast("int"))
        .show()
)
```

```python
from pyspark.sql.functions import coalesce, col, lit

(
    df_employees
        .withColumn("bonus_changed", coalesce(col("bonus").try_cast("int"), lit(0)))
        .show()
)
```

```python
(
    df_employees
        .withColumn("payment", col("salary") + coalesce(col("salary").try_cast("int"), lit(0)) / 100)
        .show()
)
```

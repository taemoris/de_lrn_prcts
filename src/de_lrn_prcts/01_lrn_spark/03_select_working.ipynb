```python
from pyspark.sql.functions import col

df_customers.select(col("id"), "first_name", "last_name").show()
```

```python
from pyspark.sql.functions import col, concat, lit


df_customers.select(
    col("id").alias("user_id"),
    "first_name",
    "last_name",
    concat(col("first_name"), lit(" "), col("last_name")).alias("fullname")
).show()
```

```python
fullname_column = concat(col("first_name"), lit(" "), col("last_name"))
fullname_column_alias = fullname_column.alias("fullname")

df_customers.select("id", fullname_column_alias).show()
```
